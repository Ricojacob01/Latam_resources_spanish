## Paso 1 — Abrir el Pipeline Editor

1. Vaya a **Jobs & Pipelines**
2. Abra su pipeline: `BNCR Medallion Pipeline - <usuario>`
3. O cree uno nuevo apuntando a la carpeta `04 - Lakeflow Declarative Pipelines/transformations/`

## Paso 2 — Abrir Genie Code

1. En la esquina **superior derecha**, clic en **Genie** / **Genie Code**
2. En la esquina **inferior derecha** del panel, active **Agent**
3. Genie Code ahora puede editar archivos SQL y ejecutar el pipeline con su aprobación

## Paso 3 — Prompt sugerido (copiar y pegar)

Use este prompt en español:

```
Estoy en el catálogo BNS del Banco Nacional de Costa Rica.
Tengo tablas bronze/silver/gold de transacciones bancarias.
Agrega una vista materializada gold llamada resumen_por_canal que muestre
por fecha y canal: total de transacciones y monto total en CRC.
Usa expectativas de calidad básicas. Crea el SQL en un archivo nuevo.
```

## Paso 4 — Revisar y aprobar

1. Genie Code mostrará un **plan** → revise y apruebe
2. Revisará el **diff** del archivo SQL propuesto
3. Apruebe la **ejecución** del pipeline
4. Verifique que la tabla se creó correctamente


````python
# Paso 5 — Verificar resultado (después de que Genie Code ejecute)
try:
    display(spark.sql(f"""
      SELECT * FROM {catalog_name}.gold.resumen_por_canal
      ORDER BY monto_total DESC
      LIMIT 10
    """))
except Exception as e:
    print("Ejecute primero Genie Code con el prompt del paso 3.")
    print(f"Detalle: {e}")
````
