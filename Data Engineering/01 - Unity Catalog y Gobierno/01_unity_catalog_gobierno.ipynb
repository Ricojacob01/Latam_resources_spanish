## A. Explorar el Catálogo BNS


```python
display(spark.sql(f"SHOW SCHEMAS IN {catalog_name}"))
display(spark.sql(f"DESCRIBE CATALOG EXTENDED {catalog_name}"))
```

## B. Etiquetas (Tags) de Gobierno


```python
spark.sql(f"ALTER SCHEMA {catalog_name}.{schema_bronze} SET TAGS ('dominio' = 'banca', 'capa' = 'bronze', 'entidad' = 'BNCR')")
spark.sql(f"ALTER SCHEMA {catalog_name}.{schema_silver} SET TAGS ('dominio' = 'banca', 'capa' = 'silver', 'entidad' = 'BNCR')")
spark.sql(f"ALTER SCHEMA {catalog_name}.{schema_gold} SET TAGS ('dominio' = 'banca', 'capa' = 'gold', 'entidad' = 'BNCR')")
print("Tags aplicados")
```
