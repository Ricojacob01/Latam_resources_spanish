```python
checks = []

def ok(msg):
    checks.append(("✅", msg))
    print("✅", msg)

def fail(msg, e):
    checks.append(("❌", f"{msg}: {e}"))
    print("❌", msg, "-", e)

ok("Variables definidas")

try:
    spark.sql("DESCRIBE CATALOG BNS")
    ok("Catálogo BNS existe")
except Exception as e:
    fail("Catálogo BNS (ejecute 00_setup)", e)

try:
    n = len(dbutils.fs.ls("/Volumes/BNS/raw/transacciones"))
    if n >= 3:
        ok(f"Volumen con {n} carpetas/archivos")
    else:
        fail("Volumen datos", f"solo {n} items — ejecute carga_datos('initial')")
except Exception as e:
    fail("Volumen", e)

try:
    spark.sql("SELECT 1").collect()
    ok("Spark SQL funciona")
except Exception as e:
    fail("Spark", e)

print("\n=== RESUMEN ===")
for s, m in checks:
    print(s, m)
```
