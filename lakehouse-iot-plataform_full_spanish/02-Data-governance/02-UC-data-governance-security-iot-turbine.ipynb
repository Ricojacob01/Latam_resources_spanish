# Implementando una gobernanza y seguridad de datos global con Unity Catalog

<img style="float: right; margin-top: 30px" width="500px" src="https://raw.githubusercontent.com/databricks-demos/dbdemos-resources/refs/heads/main/images/manufacturing/lakehouse-iot-turbine/team_flow_emily.png" />

Veamos cómo el Lakehouse puede resolver este desafío aprovechando Unity Catalog.

Nuestros datos han sido guardados como tablas Delta por el equipo de Ingeniería de Datos. El siguiente paso es asegurar estos datos permitiendo el acceso entre equipos. <br>
Una configuración típica sería:

* Ingenieros de Datos / Jobs pueden leer y actualizar los datos/esquemas principales (parte ETL)
* Científicos de Datos pueden leer las tablas finales y actualizar sus tablas de features
* Analistas de Datos tienen acceso de LECTURA a las tablas de Ingeniería de Datos y de features y pueden ingerir/transformar datos adicionales en un esquema separado
* Los datos se enmascaran/anonimizan dinámicamente según el nivel de acceso de cada usuario

Esto es posible con Unity Catalog. Cuando las tablas se guardan en Unity Catalog, pueden ponerse a disposición de toda la organización, entre workspaces y usuarios.

Unity Catalog es clave para la gobernanza de datos, incluida la creación de data products u organizar equipos alrededor de un datamesh. Entre otras cosas aporta:

* ACL de grano fino
* Audit logs
* Data lineage
* Exploración y descubrimiento de datos
* Compartir datos con organizaciones externas (Delta Sharing)


## Filtrado dinámico de datos según el usuario actual; filtrado a nivel de filas y columnas

Veamos cómo Unity Catalog puede usarse para filtrar datos y devolver resultados distintos según quién haga la consulta.

Supongamos que estamos en Chicago, y queremos que la tabla `parts` devuelva solo las piezas disponibles en la ubicación de Chicago, que es donde operamos.

Agregaremos una nueva tabla que relacione usuarios y ubicaciones de piezas *(Nota: esto también podría hacerse con grupos)*.

Tú estarás en Chicago, John en Honolulu y Lea en Denver:

```sql
-- create the table matchying the users and the country/location
CREATE OR REPLACE TABLE parts_users_country_permission (email STRING, country STRING);

INSERT INTO parts_users_country_permission (email, country)
  VALUES 
    (current_user(), 'America/Chicago'),
    ('john@mycompany.com', 'America/Honolulu'),
    ('lea@mycompany.com', 'America/Denver');
```

```sql
CREATE OR REPLACE VIEW parts_secured AS
SELECT
  CASE 
    WHEN is_account_group_member('iot_admin') THEN EAN  -- allow admin to see all
    ELSE '***' -- filter other users, they won't be able to see the EAN
  END as EAN,
  p.* EXCEPT (EAN)
FROM parts p 
INNER JOIN parts_users_country_permission u -- Get the country/location permission table
  ON p.stock_location = u.country 
  AND (u.email = current_user() OR is_account_group_member('iot_admin')); --Filter based on the current user, admin also have all permission


-- Let's test our secured view. We'll only see the 'America/Chicago' parts, and the EAN will be filtered.
SELECT * FROM parts_secured;
```


## Compartiendo datos con organizaciones externas

Hemos visto cómo otorgar (GRANT) acceso a nuestras tablas internamente (a cualquier entidad dentro de tu cuenta de Databricks).

Sin embargo, esto puede no ser suficiente. Tendrás que compartir datos con organizaciones externas (para monetización, partners, etc.).

El Lakehouse ofrece esta capacidad siendo agnóstico respecto al stack o nube de tus partners.

Esto está impulsado por [Delta Sharing](https://www.databricks.com/en/product/delta-sharing), un protocolo abierto que crea estándares abiertos y normalización de datos entre industrias.

Para un ejemplo completo de Delta Sharing, ejecuta: `dbdemos.intall('delta-sharing-airlines')`


## Más sobre gobernanza y seguridad de datos

Al reunir todos tus activos de datos, Unity Catalog te permite construir una gobernanza completa y sencilla para escalar equipos.

Unity Catalog puede usarse desde simples GRANT hasta construir una organización completa estilo datamesh.

<img src="https://github.com/QuentinAmbard/databricks-demo/raw/main/product_demos/uc/lineage/lineage-table.gif" style="float: right; margin-left: 10px"/>

### ACL de grano fino

¿Necesitas control más avanzado? Puedes elegir cambiar dinámicamente la salida de tus tablas según permisos del usuario: `dbdemos.intall('uc-01-acl')`

### Ubicaciones externas seguras (S3/ADLS/GCS)

Unity Catalog permite asegurar tus tablas gestionadas y también ubicaciones externas:  `dbdemos.intall('uc-02-external-location')`

### Lineage 

UC captura automáticamente dependencias de tablas y permite rastrear cómo se usan tus datos, incluso a nivel de fila: `dbdemos.intall('uc-03-data-lineage')`

Esto permite analizar impacto downstream o monitorear información sensible en toda la organización (GDPR).


### Audit logs

UC captura todos los eventos. ¿Necesitas saber quién accede a qué datos? Consulta tus audit logs:  `dbdemos.intall('uc-04-audit-log')`

Esto permite analizar impacto downstream o monitorear información sensible en toda la organización (GDPR).

### Migración a UC

¿Ya usas Databricks sin UC? Actualizar tus tablas para beneficiarte de Unity Catalog es simple:  `dbdemos.intall('uc-05-upgrade')`
