# Despliegue y orquestación del flujo completo

<img style="float: right; margin-left: 10px" width="600px" src="https://raw.githubusercontent.com/databricks-demos/dbdemos-resources/refs/heads/main/images/manufacturing/lakehouse-iot-turbine/di_platform_5.png" />

Todos nuestros assets están listos. Ahora debemos definir cuándo queremos que nuestro pipeline SDP se active y actualice las tablas.

Una opción es cambiar el pipeline SDP a modo continuo para tener un pipeline en streaming que provea insights casi en tiempo real.

Otra alternativa es activar el pipeline SDP cada X horas, ingerir los datos nuevos (incremental) y apagar el cómputo.

Esta es una configuración simple que ofrece un equilibrio entre tiempo en línea y latencia de ingesta.

En nuestro caso, decidimos que el mejor equilibrio es ingerir datos cada hora:

- Iniciar el pipeline SDP para ingerir datos nuevos y refrescar tablas
- Refrescar el dashboard de DBSQL (y potencialmente notificar aplicaciones downstream)
- Reentrenar nuestro modelo para incluir los datos más recientes y capturar cambios de comportamiento

## Orquestando nuestro pipeline con Databricks Workflows

<img style="float: right; margin-left: 10px" width="600px" src="https://github.com/databricks-demos/dbdemos-resources/blob/main/images/retail/lakehouse-churn/lakehouse-retail-c360-workflow.png?raw=true" />

Con la plataforma Databricks Data Intelligence, no necesitas un orquestador externo. Podemos usar [Workflows](/#job/list) (disponible en el menú izquierdo) para orquestar nuestro pipeline en pocos clics.



### Orquesta cualquier cosa en cualquier lugar
Con Workflows, puedes ejecutar cargas diversas para todo el ciclo de vida de datos y AI en cualquier nube. Orquesta Spark Declarative Pipelines y Jobs para SQL, Spark, notebooks, dbt, modelos de ML y más.

### Simple - Totalmente gestionado
Elimina la sobrecarga operativa con un servicio de orquestación totalmente gestionado, para que te enfoques en tus flujos y no en la infraestructura.

### Fiabilidad probada
Ten plena confianza en tus flujos aprovechando nuestra experiencia ejecutando decenas de millones de cargas en producción a diario en AWS, Azure y GCP.

## Creando tu workflow

<img style="float: right; margin-left: 10px" width="600px" src="https://github.com/databricks-demos/dbdemos-resources/blob/main/images/fsi/smart-claims/fsi-claims-dashboard-full.png?raw=trueg" />

Un Workflow de Databricks se compone de tareas.

Cada tarea puede ejecutar un job específico:

* Spark Declarative Pipelines
* Consulta SQL / dashboard
* Reentrenamiento / inferencia de modelos
* Notebooks
* dbt
* ...

En este ejemplo, vemos 3 tareas:

* Iniciar el pipeline SDP para ingerir nuevos datos y refrescar tablas
* Refrescar el dashboard de DBSQL (y potencialmente notificar aplicaciones downstream)
* Reentrenar nuestro modelo
