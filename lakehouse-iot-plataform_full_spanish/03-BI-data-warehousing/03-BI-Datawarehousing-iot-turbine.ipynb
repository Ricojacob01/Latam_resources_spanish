## Databricks SQL Warehouses: motor de BI de clase mundial

<img style="float: right; margin-left: 10px" width="600px" src="https://www.databricks.com/wp-content/uploads/2022/06/how-does-it-work-image-5.svg" />

Databricks SQL es un motor warehouse con miles de optimizaciones para ofrecer el mejor rendimiento para tus herramientas, tipos de consulta y aplicaciones reales. <a href='https://www.databricks.com/blog/2021/11/02/databricks-sets-official-data-warehousing-performance-record.html'>Ganó el récord oficial de rendimiento en Data Warehousing.</a>

Esto incluye el motor de consultas vectorizado de nueva generación Photon, que junto con los SQL Warehouses ofrece hasta 12x mejor precio/rendimiento que otros data warehouses en la nube.

Los **Serverless warehouse** proporcionan cómputo SQL instantáneo y elástico — desacoplado del almacenamiento — y escalan automáticamente para proveer concurrencia ilimitada sin interrupciones, ideal para casos de alta concurrencia.

Sin compromisos. Tu mejor Data Warehouse es un Lakehouse.

### Modelado de datos

Modelado integral. Guarda tus datos según tus necesidades: Data Vault, esquema en estrella, Inmon...

Databricks te permite crear PK/FK, columnas identidad (autoincrementales): `dbdemos.install('identity-pk-fk')`

### Ingesta de datos fácil con DBSQL y DBT

Capacidades llave en mano permiten a analistas e ingenieros analíticos ingerir datos fácilmente desde almacenamiento en la nube o aplicaciones empresariales como Salesforce, Google Analytics o Marketo usando Fivetran. A un clic.

Luego, gestiona dependencias y transforma datos in-place con capacidades ETL integradas en la Data Intelligence Platform (Spark Declarative Pipelines), o usando tus herramientas favoritas como dbt en Databricks SQL para el mejor rendimiento.

### Federaciones de consultas

¿Necesitas acceder a datos entre sistemas? La federación de consultas de Databricks SQL permite definir fuentes externas (ej: PostgreSQL)

### Vistas materializadas

Evita consultas costosas y materializa tus tablas. El motor recomputará solo lo necesario cuando tus datos se actualicen.
