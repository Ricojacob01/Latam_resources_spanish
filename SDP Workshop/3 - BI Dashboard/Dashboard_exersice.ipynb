# Ejercicio: Análisis de Clientes y Pedidos

## Descripción del Ejercicio

En este ejercicio, crearás un dashboard interactivo para analizar datos de clientes y pedidos. El objetivo es que explores la base de datos, generes visualizaciones significativas y respondas preguntas clave sobre el negocio.

## Instrucciones

1. **Explora los datos disponibles**: Familiarízate con las tablas de clientes y pedidos en la base de datos
2. **Selecciona 3 preguntas**: De las 6 preguntas listadas abajo, elige 3 que te parezcan más relevantes para el análisis
3. **Escribe el código**: Crea celdas de código para responder cada pregunta seleccionada
4. **Genera visualizaciones**: Utiliza gráficos apropiados (barras, líneas, tablas) para presentar tus resultados
5. **Crea un dashboard**: Organiza tus visualizaciones en un dashboard coherente y fácil de interpretar

## Preguntas para Responder (Selecciona 3)

1. **¿Qué ciudad tiene el mayor número de clientes?**
   - Identifica la ciudad con la mayor concentración de clientes

2. **¿Cuántos clientes únicos hay en la base de datos?**
   - Calcula el total de clientes únicos registrados

3. **¿Cuál es el número total de pedidos realizados en el día más reciente?**
   - Encuentra la fecha más reciente y cuenta los pedidos de ese día

4. **¿Qué estados tienen más clientes?**
   - Analiza la distribución de clientes por estado (top 5 o top 10)

5. **¿Cuántos clientes hay en cada ciudad?**
   - Genera un conteo completo de clientes por ciudad

6. **¿En qué fecha se registró el mayor número de pedidos diarios?**
   - Identifica el día con el pico más alto de pedidos

## Consejos

* Utiliza SQL o PySpark según tu preferencia
* Asegúrate de que tus visualizaciones sean claras y tengan títulos descriptivos
* Incluye comentarios en tu código para explicar tu lógica
* Verifica que los datos no contengan valores nulos que puedan afectar tus resultados

¡Buena suerte con tu análisis!
