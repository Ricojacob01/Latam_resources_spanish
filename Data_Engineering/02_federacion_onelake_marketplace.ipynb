## C. Fuentes que se pueden federar con Lakehouse Federation

| Fuente | Conector | Notas |
|--------|----------|-------|
| **MySQL** | JDBC | Bases de datos transaccionales |
| **PostgreSQL** | JDBC | Incluye Aurora, Cloud SQL, Azure DB |
| **SQL Server** | JDBC | Incluye Azure SQL y AWS RDS |
| **Oracle** | JDBC | Versiones 11g+ |
| **Amazon Redshift** | JDBC | Data warehousing en AWS |
| **Snowflake** | Conector nativo | Cross-cloud |
| **Google BigQuery** | Conector nativo | Analytics en GCP |
| **Databricks** | Unity Catalog | Workspace a workspace |
| **Microsoft Fabric / OneLake** | Shortcuts + mirroring | Integración bidireccional |
| **Teradata** | JDBC | Enterprise data warehousing |
| **SAP HANA** | JDBC | Datos empresariales SAP |
| **Netsuite** | JDBC | ERP/CRM cloud |
| **Salesforce** | Lakeflow Connect | CRM cloud |
| **Delta Sharing** | Protocolo abierto | Compartir datos entre organizaciones sin copiar |

> **Ventaja clave:** Consultas SQL estándar sobre fuentes externas sin mover datos. Unity Catalog gobierna permisos y linaje de forma centralizada.
